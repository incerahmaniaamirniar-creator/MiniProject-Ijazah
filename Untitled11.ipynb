```python
from google.colab import files

# 1. Unggah ke-9 gambar ijazah
print("Unggah 9 file ijazah sekaligus:")
uploaded = files.upload()

# 2. Tampung daftar nama filenya
files_ijazah = sorted(list(uploaded.keys()))

# 3. Buka dan tampilkan hasil Grayscale untuk ke-9 ijazah
plt.figure(figsize=(15, 12))

for idx, file_path in enumerate(files_ijazah):
    img = cv2.imread(file_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    plt.subplot(3, 3, idx + 1)
    plt.title(f"Grayscale: {file_path[:15]}...", fontsize=9)
    plt.imshow(gray, cmap='gray')
    plt.axis('off')

plt.tight_layout()
plt.show()

print("Langkah 2 Selesai: 9 Ijazah berhasil diunggah dan diubah ke Grayscale!")
```

```python
# 1. Objek CLAHE (Contrast Limited Adaptive Histogram Equalization)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

# 2. Terapkan CLAHE & tampilkan visualisasinya untuk 9 ijazah
plt.figure(figsize=(15, 12))

for idx, file_path in enumerate(files_ijazah):
    img = cv2.imread(file_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    enhanced = clahe.apply(gray)

    plt.subplot(3, 3, idx + 1)
    plt.title(f"Enhanced (CLAHE): {file_path[:15]}...", fontsize=9)
    plt.imshow(enhanced, cmap='gray')
    plt.axis('off')

plt.tight_layout()
plt.show()

print("Langkah 3 Selesai: Image Enhancement (CLAHE) berhasil diterapkan pada 9 ijazah!")
```

```python
# Tampilkan potongan area nomor dan tanda tangan untuk ke-9 ijazah
plt.figure(figsize=(15, 12))

for idx, file_path in enumerate(files_ijazah):
    img = cv2.imread(file_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    enhanced = clahe.apply(gray)
    h, w = gray.shape

    # Crop Area Nomor & Tanda Tangan
    roi_nomor = enhanced[int(h * 0.88):int(h * 0.96), int(w * 0.04):int(w * 0.35)]
    roi_ttd = gray[int(h * 0.70):int(h * 0.82), int(w * 0.12):int(w * 0.40)]

    # Plot Area Nomor
    plt.subplot(9, 2, (idx * 2) + 1)
    plt.title(f"ROI Nomor: {file_path[:12]}...", fontsize=8)
    plt.imshow(roi_nomor, cmap='gray')
    plt.axis('off')

    # Plot Area Tanda Tangan
    plt.subplot(9, 2, (idx * 2) + 2)
    plt.title("ROI Tanda Tangan", fontsize=8)
    plt.imshow(roi_ttd, cmap='gray')
    plt.axis('off')

plt.tight_layout()
plt.show()

print("Langkah 4 Selesai: Pemotongan area (ROI) untuk 9 ijazah berhasil dilakukan!")
```

```python
plt.figure(figsize=(15, 12))

for idx, file_path in enumerate(files_ijazah):
    img = cv2.imread(file_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    enhanced = clahe.apply(gray)
    h, w = gray.shape

    # 1. Crop Area ROI
    roi_nomor = enhanced[int(h * 0.88):int(h * 0.96), int(w * 0.04):int(w * 0.35)]
    roi_ttd = gray[int(h * 0.70):int(h * 0.82), int(w * 0.12):int(w * 0.40)]

    # 2. Thresholding & OCR Nomor
    _, roi_nomor_thresh = cv2.threshold(roi_nomor, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    raw_ocr = pytesseract.image_to_string(roi_nomor_thresh, config='--oem 3 --psm 6')
    nomor_ijazah = raw_ocr.strip()
    if "Nomor ijazah:" in nomor_ijazah:
        nomor_ijazah = nomor_ijazah.replace("Nomor ijazah:", "").strip()

    # 3. Thresholding & Morfologi Tanda Tangan
    _, roi_ttd_thresh = cv2.threshold(roi_ttd, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))
    roi_ttd_morph = cv2.morphologyEx(roi_ttd_thresh, cv2.MORPH_CLOSE, kernel)

    rasio_tinta = cv2.countNonZero(roi_ttd_morph) / (roi_ttd_morph.shape[0] * roi_ttd_morph.shape[1])
    status_ttd = "PRESENT" if rasio_tinta > 0.01 else "ABSENT"

    # 4. Plot Visualisasi Hasil Binarisasi & Morfologi
    plt.subplot(9, 2, (idx * 2) + 1)
    plt.title(f"Binarisasi Nomor: {file_path[:12]}...", fontsize=8)
    plt.imshow(roi_nomor_thresh, cmap='gray')
    plt.axis('off')

    plt.subplot(9, 2, (idx * 2) + 2)
    plt.title(f"Morfologi TTD ({status_ttd})", fontsize=8)
    plt.imshow(roi_ttd_morph, cmap='gray')
    plt.axis('off')

    # 5. Cetak Output Teks Sesuai Format Dosen
    print("----------------------------------------")
    print(f"Input:\n{file_path}\n")
    print("Output:")
    print(f"Nomor Ijazah : {nomor_ijazah}")
    print(f"Tanda Tangan : {status_ttd}")

plt.tight_layout()
plt.show()

print("\nLangkah 5 Selesai: Ekstraksi OCR dan Deteksi Tanda Tangan Berhasil!")
```

```python
import pandas as pd
import Levenshtein

# Ground Truth (Nomor ijazah asli yang seharusnya)
ground_truth = "571012022000056"

# Fungsi menghitung nilai CER
def hitung_cer(ref, hyp):
    digits_only = re.sub(r'\D', '', hyp)
    if len(digits_only) == 0:
        return 1.0
    dist = Levenshtein.distance(ref, digits_only)
    return dist / len(ref)

hasil_evaluasi = []

# Loop pengujian nilai CER pada ke-9 ijazah
for file_path in files_ijazah:
    if not os.path.exists(file_path): continue

    img = cv2.imread(file_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    h, w = gray.shape
    roi = gray[int(h*0.88):int(h*0.96), int(w*0.04):int(w*0.35)]

    # 1. Tanpa Enhancement (Grayscale Saja)
    _, t1 = cv2.threshold(roi, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    ocr1 = pytesseract.image_to_string(t1, config='--oem 3 --psm 6')
    cer1 = hitung_cer(ground_truth, ocr1)

    # 2. CLAHE (Adaptive Enhancement)
    clahe_obj = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    roi_clahe = clahe_obj.apply(roi)
    _, t2 = cv2.threshold(roi_clahe, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    ocr2 = pytesseract.image_to_string(t2, config='--oem 3 --psm 6')
    cer2 = hitung_cer(ground_truth, ocr2)

    # 3. Histogram Equalization Global
    roi_he = cv2.equalizeHist(roi)
    _, t3 = cv2.threshold(roi_he, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    ocr3 = pytesseract.image_to_string(t3, config='--oem 3 --psm 6')
    cer3 = hitung_cer(ground_truth, ocr3)

    hasil_evaluasi.append({
        'Nama File': file_path[:15] + "...",
        'CER Grayscale': f"{cer1*100:.1f}%",
        'CER CLAHE': f"{cer2*100:.1f}%",
        'CER HistEq Global': f"{cer3*100:.1f}%"
    })

# Tampilkan Tabel
df_hasil = pd.DataFrame(hasil_evaluasi)
print("=" * 65)
print("TABEL EVALUASI NILAI CER (CHARACTER ERROR RATE)")
print("=" * 65)
print(df_hasil.to_string(index=False))
```
